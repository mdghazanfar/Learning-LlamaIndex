```python
import os
from dotenv import load_dotenv
from llama_index.llms.openai import OpenAI
```

```python
load_dotenv()
api_key = os.getenv("OPEN_API_KEY")
```

```python
llm = OpenAI(
    model="gpt-4o-mini",
    api_key=api_key,
)
```

```python
response = llm.stream_complete("What is the llama-index?")
for chunk in response:
    print(chunk.delta, end='') # try excluding delta parameter to see the full response
```

```python
# complete - Generate completion for the given prompt.
# stream_complete - Generate completion for the given prompt in a streaming manner.
# Delta received from the stream chat. Delta means the change in the response since the last chunk.
# The delta can be a text, image, or tool call.
```
