```python
import os
from dotenv import load_dotenv
from llama_index.embeddings.openai import OpenAIEmbedding
from llama_index.core import SimpleDirectoryReader
from llama_index.core import VectorStoreIndex
from llama_index.core import StorageContext
from llama_index.core import load_index_from_storage
```

```python
load_dotenv()
api_key = os.getenv("OPEN_API_KEY")
```

```python
openai_embedding = OpenAIEmbedding(model="text-embedding-3-small", api_key=api_key)
```

```python
reader = SimpleDirectoryReader(input_dir="data", required_exts=[".txt"])
documents = reader.load_data()

print(f"Documents: {documents}")
```

```python
# reader = SimpleDirectoryReader(input_dir="data", recursive=True, required_exts=[".txt"]) # use recursive=True to read all files in subdirectories
```

```python
# store the documents in a vector store index
index = VectorStoreIndex.from_documents(documents=documents)

print(f"Vectors Store: {index}")
print(f"Vectors Store ID: {index.index_id}")
```

```python
# perform the embeddings for the documents in the vector store
print(f"Number of documents in the vector store: {len(index.docstore.docs)}")
```
