```python
import os
from dotenv import load_dotenv
from llama_index.embeddings.google_genai import GoogleGenAIEmbedding
```

```python
load_dotenv()
api_key = os.getenv("GOOGLE_API_KEY")
```

```python
gemini_embedding = GoogleGenAIEmbedding(
    model_name="models/text-embedding-004",
    api_key=api_key,
)
```

```python
text = "LLMs does the reasoning"
embed_vector = gemini_embedding.get_text_embedding(text)

print(f"Embedding dimensions: {len(embed_vector)}")
print(f"First 5 values: {embed_vector[:5]}")
```
